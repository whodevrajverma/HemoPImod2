## Section 04 | Prediction Models

HemoPI-MOD2 supports two prediction approaches:

- **Model 1 — Structure-based prediction**
- **Model 2 — SMILES-based prediction**

FASTA sequences can be processed using either approach. For Model 2, MAP sequences are internally converted into their corresponding SMILES representations before prediction.

```python
! hemopimod2 -i input.fasta -j 1 -m 1 -o outfile1.csv
```

```python
! hemopimod2 -i input.fasta -j 1 -m 2 -o outfile2.csv
```

## Section 05 | Direct SMILES Input

Alternatively, Model 2 accepts SMILES representations directly, allowing users to bypass the internal sequence-to-SMILES conversion step.

Provide the input structures in the `input_smiles.txt` format. This option is particularly useful for independent validation and comparative prediction.

```python
! hemopimod2 -i input_smiles.txt -j 1 -m 2 -o outfile3.csv
```

## Section 06 | Structure-Based Prediction (Model 3)

Example PDB files have already been retrieved from the HemoPI-MOD2 GitHub repository in the previous step.

Model 3 supports two modes of input:

- **Single structure:** Provide the complete path to an individual `.pdb` file for prediction.
- **Batch processing:** Provide the path to a directory containing multiple `.pdb` files to process all available structures together.

The model extracts 100 Mordred 2D descriptors from the supplied structures and uses the trained machine learning model to predict their hemolytic potency (HC50).

```python
! hemopimod2 -i input.pdb -j 1 -m 3 -o outfile4.csv
```

## Section 07 | Peptide Design – Single-Residue Mutagenesis

The Design module `-j 2` enables systematic exploration of single-residue substitutions to identify potentially improved peptide analogues.

Provide the lead peptide sequence in MAP/FASTA format. The module generates all possible single-residue substitutions at each eligible position and evaluates each variant using the Sequence-based prediction model (Model 1).

The output includes predicted HC50 values, fold changes relative to the wild-type peptide, and the direction of change in hemolytic potency.

**Note:** Chemical modifications are retained during mutagenesis. Only standard amino acid residues are substituted. The Design module supports peptides of up to 50 residues.

**Filtering Large Libraries:** Because single-point saturation mutagenesis generates large tables, combine `-j 2` with `-d 1` if you only want to retain hemolytic variants.

```python
! hemopimod2 -i input.fasta -j 2 -m 1 -o outfile5.csv
```

## Section 08 | Result Analysis and Interpretation

Upon completion, design predictions are generated in CSV format for further analysis.

The output provides:
- **pHC50:** Predicted logarithmic hemolysis concentration.
- **Prediction HC50 (µM):** Predicted concentration required to cause 50% hemolysis.
- **Fold Change:** Relative change in HC50 compared with the wild-type peptide (Design module).
- **Hemolytic Activity:** Classification based on the established 100 µM threshold.

A higher HC50 indicates lower hemolytic activity, whereas a lower HC50 indicates greater hemolytic activity.

The generated CSV file contains the complete prediction results and can be downloaded for downstream analysis.

```python
import csv
from IPython.display import HTML, display

with open('outfile5.csv', 'r', newline='') as f:
  rows = [next(csv.reader(f)) for _ in range(6)]  # Header + 5 rows

html = '<table border="1" style="border-collapse: collapse; text-align: left;">'
html += (
    '<thead><tr>'
    + ''.join(f'<th style="padding: 6px;">{c}</th>' for c in rows[0])
    + '</tr></thead><tbody>'
)
for row in rows[1:]:
  html += (
      '<tr>'
      + ''.join(f'<td style="padding: 6px;">{c}</td>' for c in row)
      + '</tr>'
  )
html += '</tbody></table>'

display(HTML(html))
```

## Section 09 | Summary, Quick Reference & Downloading Results

You have now completed the end-to-end workflow for **HemoPI-MOD2**—from environment setup and sequence/chemical predictions (`-m 1`, `-m 2`) to 3D structure-based scoring (`-m 3`) and single-point mutant library design (`-j 2`).

### Quick CLI Reference Cheatsheet

| Flag | Argument | Description |
| :--- | :--- | :--- |
| **`-i`** | `<file>` | Input file path (`.fasta`, `.txt`/SMILES, or `.pdb`). |
| **`-j`** | `1` \| `2` | Job mode: `1` = Prediction (default), `2` = Design/Analog generation (Model 1 only). |
| **`-m`** | `1` \| `2` \| `3` | Prediction model: `1` = Sequence (MAP/FASTA), `2` = Chemical (SMILES), `3` = Structure (PDB). |
| **`-d`** | `1` \| `2` | Display filter: `1` = Hemolytic peptides only, `2` = All peptides (default). |
| **`-o`** | `<file>` | Output CSV filename (default: `outfile.csv`). |
| **`-wd`** | `<dir>` | Custom working directory for reading/saving files. |
